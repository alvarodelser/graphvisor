```python
from neo4j import GraphDatabase

driver = GraphDatabase.driver(
    "bolt+s://wiig.dia.fi.upm.es:8474",
    auth=("neo4j", "password")
)

query = """
MATCH (e:Entity)<-[:HAS_SUBJECT|HAS_OBJECT]-(a)
WITH e, count(a) as degree
ORDER BY degree DESC
LIMIT 10
RETURN e.name AS Entity name, e.reasoning AS reasoning, e.confidence AS confidence, e.source_argument_id as arg_id
"""

with driver.session() as session:
    result = session.run(query)

    for record in result:
        print(record)
```
